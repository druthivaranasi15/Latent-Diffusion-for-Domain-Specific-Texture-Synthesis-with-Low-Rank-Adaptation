```python
#7
import hashlib
from collections import defaultdict
from PIL import Image

image_records = []
corrupted_images = []
hash_groups = defaultdict(list)

for class_name in SELECTED_CLASSES:

    folder = CLASS_PATHS[class_name]

    for image_path in sorted(folder.iterdir()):

        if image_path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        try:
            with Image.open(image_path) as img:
                img.verify()

            file_hash = hashlib.sha256(
                image_path.read_bytes()
            ).hexdigest()

            record = {
                "path": str(image_path),
                "class": class_name,
                "hash": file_hash
            }

            image_records.append(record)
            hash_groups[file_hash].append(record)

        except Exception as e:
            corrupted_images.append({
                "path": str(image_path),
                "error": str(e)
            })

duplicate_groups = {
    h: records
    for h, records in hash_groups.items()
    if len(records) > 1
}

print("Total images scanned:", 
      len(image_records) + len(corrupted_images))

print("Valid images:", len(image_records))
print("Corrupted images:", len(corrupted_images))
print("Exact duplicate groups:", len(duplicate_groups))
print(
    "Images involved in duplicates:",
    sum(len(v) for v in duplicate_groups.values())
)
```

```python
#8
from collections import defaultdict
import pandas as pd

hash_to_records = defaultdict(list)

for record in image_records:
    hash_to_records[record["hash"]].append(record)

clean_records = []
ambiguous_groups = 0
removed_duplicates = 0

for file_hash, records in hash_to_records.items():

    labels = set(r["class"] for r in records)

    if len(labels) > 1:
        ambiguous_groups += 1
        continue

    records = sorted(records, key=lambda r: r["path"])

    clean_records.append(records[0])
    removed_duplicates += len(records) - 1

print("Valid images:", len(image_records))
print("Removed duplicates:", removed_duplicates)
print("Ambiguous duplicate groups:", ambiguous_groups)
print("Final clean images:", len(clean_records))
```

```python
#9
from sklearn.model_selection import train_test_split

df = pd.DataFrame(clean_records)

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df["class"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["class"]
)

train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

train_df["split"] = "train"
val_df["split"] = "validation"
test_df["split"] = "test"

final_df = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

final_df.to_csv(
    OUTPUT_ROOT / "dataset_manifest.csv",
    index=False
)

print("Training images:", len(train_df))
print("Validation images:", len(val_df))
print("Testing images:", len(test_df))
print("Total images:", len(final_df))

print("\nClass-wise distribution:")
print(
    final_df.groupby(["class", "split"])
    .size()
    .unstack(fill_value=0)
)
```

```python
#17

PROMPTS = {
    "Tomato_Bacterial_spot":
        "A realistic close-up botanical photograph of a tomato leaf affected by bacterial spot disease, showing small dark lesions on the leaf surface, natural lighting, detailed leaf texture",

    "Tomato_Early_blight":
        "A realistic close-up botanical photograph of a tomato leaf affected by early blight disease, showing characteristic brown concentric lesions and yellowing, natural lighting, detailed leaf texture",

    "Tomato_Late_blight":
        "A realistic close-up botanical photograph of a tomato leaf affected by late blight disease, showing irregular dark brown lesions and damaged leaf tissue, natural lighting, detailed leaf texture",

    "Tomato_healthy":
        "A realistic close-up botanical photograph of a healthy green tomato leaf, showing natural veins, uniform green coloration and detailed leaf texture"
}

NEGATIVE_PROMPT = (
    "cartoon, painting, illustration, blurry, low quality, "
    "watermark, text, distorted leaves, artificial colors"
)

print("Number of classes:", len(PROMPTS))
```

```python
#18

import torch
import pandas as pd
from pathlib import Path
from tqdm.auto import tqdm

BASELINE_DIR.mkdir(parents=True, exist_ok=True)

BASE_SEED = 42
NUM_IMAGES_PER_CLASS = 100

generation_records = []

for class_index, class_name in enumerate(SELECTED_CLASSES):

    print(f"\nGenerating: {class_name}")

    class_dir = BASELINE_DIR / class_name
    class_dir.mkdir(parents=True, exist_ok=True)

    prompt = PROMPTS[class_name]

    for image_index in tqdm(range(NUM_IMAGES_PER_CLASS)):

        seed = BASE_SEED + class_index * 1000 + image_index

        image_name = f"baseline_{image_index + 1:04d}.png"
        image_path = class_dir / image_name

        # Reuse an existing image if the cell is rerun
        if image_path.exists():
            continue

        generator = torch.Generator(
            device="cuda"
        ).manual_seed(seed)

        with torch.inference_mode():

            result = pipe(
                prompt=prompt,
                negative_prompt=NEGATIVE_PROMPT,
                height=512,
                width=512,
                num_inference_steps=25,
                guidance_scale=7.5,
                generator=generator
            )

        result.images[0].save(image_path)

        generation_records.append({
            "class": class_name,
            "image_name": image_name,
            "seed": seed,
            "model": MODEL_ID,
            "method": "Baseline Stable Diffusion 1.5",
            "prompt": prompt,
            "image_path": str(image_path)
        })

        del result, generator

    torch.cuda.empty_cache()

# Reconstruct complete metadata, including previously saved images
all_records = []

for class_index, class_name in enumerate(SELECTED_CLASSES):
    class_dir = BASELINE_DIR / class_name

    for image_index in range(NUM_IMAGES_PER_CLASS):
        image_name = f"baseline_{image_index + 1:04d}.png"
        image_path = class_dir / image_name

        if image_path.exists():
            all_records.append({
                "class": class_name,
                "image_name": image_name,
                "seed": BASE_SEED + class_index * 1000 + image_index,
                "model": MODEL_ID,
                "method": "Baseline Stable Diffusion 1.5",
                "prompt": PROMPTS[class_name],
                "image_path": str(image_path)
            })

generation_df = pd.DataFrame(all_records)

generation_df.to_csv(
    BASELINE_DIR / "generation_metadata.csv",
    index=False
)

print("\nBaseline generation completed!")
print("Total generated images:", len(generation_df))
print(generation_df.groupby("class").size())
```

```python

import os
import shutil
import numpy as np
import pandas as pd
import torch
from PIL import Image
from pathlib import Path

# --------------------------------------------------
# 1. Paths and metadata
# --------------------------------------------------
metadata_path = BASELINE_DIR / "generation_metadata.csv"

# Keep a backup of the original metadata
backup_path = BASELINE_DIR / "generation_metadata_before_retry.csv"
if not backup_path.exists():
    shutil.copy2(metadata_path, backup_path)

df = pd.read_csv(metadata_path)

assert "pipe" in globals(), (
    "Stable Diffusion pipeline 'pipe' is not loaded. "
    "Run the pipeline-loading cell before this cell."
)

# Use the original negative prompt if it exists in the notebook.
negative_prompt = globals().get(
    "NEGATIVE_PROMPT",
    globals().get("negative_prompt", "")
)

# --------------------------------------------------
# 2. Identify black images again
# --------------------------------------------------
def is_black_image(image_path):
    try:
        img = np.array(Image.open(image_path).convert("RGB"))
        return img.mean() < 2 and img.std() < 3
    except Exception:
        return True

failed_images = []

for image_path in BASELINE_DIR.rglob("*.png"):
    if is_black_image(image_path):
        failed_images.append(image_path)

print("Images requiring regeneration:", len(failed_images))

# --------------------------------------------------
# 3. Regenerate only failed images
# --------------------------------------------------
MAX_RETRIES = 10
successful = 0
unsuccessful = []

for image_path in failed_images:

    class_name = image_path.parent.name
    image_name = image_path.name

    matching_rows = df.index[
        (df["class"] == class_name) &
        (df["image_name"] == image_name)
    ].tolist()

    if len(matching_rows) != 1:
        print("Metadata match problem:", image_name)
        unsuccessful.append(image_path)
        continue

    row_idx = matching_rows[0]
    original_seed = int(df.loc[row_idx, "seed"])
    prompt = str(df.loc[row_idx, "prompt"])

    print(f"\nRegenerating: {class_name}/{image_name}")

    regenerated = False

    for attempt in range(1, MAX_RETRIES + 1):

        # Deterministic replacement seed
        new_seed = original_seed + 100000 + attempt * 1009

        generator = torch.Generator(
            device="cuda"
        ).manual_seed(new_seed)

        with torch.inference_mode():
            result = pipe(
                prompt=prompt,
                negative_prompt=negative_prompt,
                height=512,
                width=512,
                num_inference_steps=25,
                guidance_scale=7.5,
                generator=generator
            )

        generated_image = result.images[0]

        nsfw_flags = getattr(
            result, "nsfw_content_detected", None
        )

        flagged = (
            nsfw_flags is not None
            and len(nsfw_flags) > 0
            and bool(nsfw_flags[0])
        )

        if flagged:
            print(f"  Attempt {attempt}: safety checker flagged image.")
            continue

        # Additional protection against saving a black image
        temp_path = image_path.with_name(
            image_path.stem + "_retry_tmp.png"
        )

        generated_image.save(temp_path)

        if is_black_image(temp_path):
            temp_path.unlink(missing_ok=True)
            print(f"  Attempt {attempt}: black image; retrying.")
            continue

        # Replace the failed image with the valid image
        os.replace(temp_path, image_path)

        # Record the replacement seed
        df.loc[row_idx, "seed"] = new_seed
        df.loc[row_idx, "image_path"] = str(image_path)

        # Save metadata after every successful replacement
        df.to_csv(metadata_path, index=False)

        successful += 1
        regenerated = True

        print(f"  Successfully regenerated using seed {new_seed}")
        break

    if not regenerated:
        unsuccessful.append(image_path)
        print("  Could not regenerate after maximum retries.")

# --------------------------------------------------
# 4. Final report
# --------------------------------------------------
print("\n========== REGENERATION SUMMARY ==========")
print("Successfully regenerated:", successful)
print("Still unsuccessful:", len(unsuccessful))

for path in unsuccessful:
    print("Needs attention:", path)

print("Updated metadata:", metadata_path)
```

```python

import pandas as pd

CLASSES = [
    "Tomato_Bacterial_spot",
    "Tomato_Early_blight",
    "Tomato_Late_blight",
    "Tomato_healthy"
]

BASELINE_METADATA = (
    PROJECT_ROOT
    / "generated_images"
    / "baseline"
    / "generation_metadata.csv"
)

if not BASELINE_METADATA.exists():
    raise FileNotFoundError(
        f"Baseline metadata is missing: {BASELINE_METADATA}"
    )

baseline_df = pd.read_csv(BASELINE_METADATA)

# Reuse the exact prompts from baseline generation
PROMPTS = {}

for class_name in CLASSES:

    class_prompts = baseline_df.loc[
        baseline_df["class"] == class_name,
        "prompt"
    ].dropna().unique()

    if len(class_prompts) == 0:
        raise ValueError(f"No baseline prompt found for {class_name}")

    if len(class_prompts) > 1:
        raise ValueError(
            f"Multiple different prompts found for {class_name}. "
            "Check the baseline metadata."
        )

    PROMPTS[class_name] = class_prompts[0]

# Same negative prompt used in baseline Cell #17
NEGATIVE_PROMPT = (
    "cartoon, painting, illustration, blurry, low quality, "
    "watermark, text, distorted leaves, artificial colors"
)

NUM_IMAGES_PER_CLASS = 100
IMAGE_SIZE = 512
INFERENCE_STEPS = 25
GUIDANCE_SCALE = 7.5

print("Baseline prompts loaded successfully!")

for class_name, prompt in PROMPTS.items():
    print(f"\n{class_name}: {prompt}")

print("\nExpected images:", len(CLASSES) * NUM_IMAGES_PER_CLASS)
print("Negative prompt:", NEGATIVE_PROMPT)
```

```python
from pathlib import Path
from PIL import Image
from tqdm.auto import tqdm
import numpy as np
import pandas as pd
import gc

NUM_IMAGES_PER_CLASS = 100
IMAGE_SIZE = 512
INFERENCE_STEPS = 25
GUIDANCE_SCALE = 7.5
MAX_RETRIES = 5

LORA_GENERATED_DIR = (
    PROJECT_ROOT / "generated_images" / "lora_rank8"
)

LORA_GENERATED_DIR.mkdir(parents=True, exist_ok=True)

def is_black_image(image):
    arr = np.asarray(image.convert("RGB"))
    return arr.mean() < 5 and arr.std() < 3

records = []

for class_index, class_name in enumerate(CLASSES):

    print(f"\nGenerating: {class_name}")

    class_dir = LORA_GENERATED_DIR / class_name
    class_dir.mkdir(parents=True, exist_ok=True)

    for image_index in tqdm(range(NUM_IMAGES_PER_CLASS)):

        image_name = f"{class_name}_{image_index + 1:04d}.png"
        image_path = class_dir / image_name

        if image_path.exists():
            raise FileExistsError(
                f"{image_path} already exists. Use a fresh output folder "
                "or verify the previous generation run."
            )

        success = False
        used_seed = None

        for attempt in range(MAX_RETRIES):

            seed = 80000 + class_index * 10000 + image_index * 10 + attempt
            generator = torch.Generator("cuda").manual_seed(seed)

            with torch.inference_mode():
                result = pipe_lora(
                    prompt=PROMPTS[class_name],
                    negative_prompt=NEGATIVE_PROMPT,
                    height=IMAGE_SIZE,
                    width=IMAGE_SIZE,
                    num_inference_steps=INFERENCE_STEPS,
                    guidance_scale=GUIDANCE_SCALE,
                    generator=generator
                )

            generated_image = result.images[0]

            flags = getattr(result, "nsfw_content_detected", None)
            flagged = (
                flags is not None
                and len(flags) > 0
                and bool(flags[0])
            )

            if flagged or is_black_image(generated_image):
                continue

            generated_image.save(image_path)
            used_seed = seed
            success = True
            break

        records.append({
            "class": class_name,
            "image_name": image_name,
            "seed": used_seed,
            "model": MODEL_ID,
            "method": "Stable_Diffusion_1.5_LoRA_Rank8",
            "prompt": PROMPTS[class_name],
            "image_path": str(image_path),
            "status": "success" if success else "failed"
        })

    gc.collect()
    torch.cuda.empty_cache()

metadata_df = pd.DataFrame(records)

metadata_path = LORA_GENERATED_DIR / "generation_metadata.csv"
metadata_df.to_csv(metadata_path, index=False)

print("\nGeneration finished.")
print(metadata_df.groupby(["class", "status"]).size())
print("Metadata saved:", metadata_path)
```

```python
import os
import json
import time
import random
import copy
import gc
import zipfile

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn

from pathlib import Path
from torch.utils.data import DataLoader, ConcatDataset
from torchvision import datasets, transforms, models
from torchvision.models import ResNet18_Weights

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from tqdm.auto import tqdm

# --------------------------------------------------
# PROJECT PATHS
# --------------------------------------------------

PROJECT_ROOT = Path("/kaggle/working/plant_disease_project")

REAL_DATASET_DIR = PROJECT_ROOT / "real_dataset"

REAL_TRAIN_DIR = REAL_DATASET_DIR / "train"
REAL_VAL_DIR = REAL_DATASET_DIR / "validation"
REAL_TEST_DIR = REAL_DATASET_DIR / "test"

BASELINE_SYNTHETIC_DIR = (
    PROJECT_ROOT / "generated_images" / "baseline"
)

LORA_SYNTHETIC_DIR = (
    PROJECT_ROOT / "generated_images" / "lora_rank8"
)

RESULTS_ROOT = PROJECT_ROOT / "results"

# --------------------------------------------------
# CLASSIFIER CONFIGURATION
# --------------------------------------------------

SEED = 42
BATCH_SIZE = 32
NUM_CLASSES = 4
IMAGE_SIZE = 224
NUM_WORKERS = 2

EPOCHS = 15
PATIENCE = 4

LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-4

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

# --------------------------------------------------
# DATA TRANSFORMS
# --------------------------------------------------

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("\nClassifier configuration ready!")
```

```python
def set_seed(seed):

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


def run_classifier_experiment(
    synthetic_dir,
    experiment_name,
    seed=42
):

    print("\n" + "=" * 60)
    print("EXPERIMENT:", experiment_name)
    print("=" * 60)

    # Prevent accidental overwriting of previous results
    RESULT_DIR = RESULTS_ROOT / experiment_name

    if RESULT_DIR.exists() and any(RESULT_DIR.iterdir()):
        raise FileExistsError(
            f"Results already exist: {RESULT_DIR}. "
            "Choose a new output folder or inspect the existing results."
        )

    MODEL_DIR = RESULT_DIR / "models"
    MODEL_DIR.mkdir(parents=True, exist_ok=True)

    # --------------------------------------------------
    # 1. LOAD DATASETS
    # --------------------------------------------------

    set_seed(seed)

    real_train = datasets.ImageFolder(
        REAL_TRAIN_DIR,
        transform=train_transform
    )

    synthetic_train = datasets.ImageFolder(
        synthetic_dir,
        transform=train_transform
    )

    real_val = datasets.ImageFolder(
        REAL_VAL_DIR,
        transform=eval_transform
    )

    real_test = datasets.ImageFolder(
        REAL_TEST_DIR,
        transform=eval_transform
    )

    assert real_train.class_to_idx == synthetic_train.class_to_idx
    assert real_train.class_to_idx == real_val.class_to_idx
    assert real_train.class_to_idx == real_test.class_to_idx

    assert len(real_train) == 4629
    assert len(synthetic_train) == 400
    assert len(real_val) == 992
    assert len(real_test) == 992

    train_dataset = ConcatDataset([
        real_train,
        synthetic_train
    ])

    # --------------------------------------------------
    # 2. DATA LOADERS
    # --------------------------------------------------

    loader_generator = torch.Generator()
    loader_generator.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        generator=loader_generator
    )

    val_loader = DataLoader(
        real_val,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )

    test_loader = DataLoader(
        real_test,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )

    print("\nDataset verification:")
    print("Real training:", len(real_train))
    print("Synthetic training:", len(synthetic_train))
    print("Combined training:", len(train_dataset))
    print("Validation:", len(real_val))
    print("Testing:", len(real_test))

    # --------------------------------------------------
    # 3. INITIALIZE RESNET-18
    # --------------------------------------------------

    weights = ResNet18_Weights.DEFAULT

    model = models.resnet18(weights=weights)

    model.fc = nn.Linear(
        model.fc.in_features,
        NUM_CLASSES
    )

    model = model.to(device)

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.5,
        patience=2
    )

    # --------------------------------------------------
    # 4. TRAINING
    # --------------------------------------------------

    best_val_accuracy = 0.0
    best_state = None

    epochs_without_improvement = 0

    history = []

    start_time = time.time()

    for epoch in range(EPOCHS):

        model.train()

        running_loss = 0.0
        correct = 0
        total = 0

        progress = tqdm(
            train_loader,
            desc=f"Epoch {epoch + 1}/{EPOCHS}"
        )

        for images, labels in progress:

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            optimizer.zero_grad(set_to_none=True)

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            optimizer.step()

            running_loss += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)

            correct += (
                predictions == labels
            ).sum().item()

            total += labels.size(0)

            progress.set_postfix(
                loss=f"{loss.item():.4f}"
            )

        train_loss = running_loss / total
        train_accuracy = correct / total

        # --------------------------------------------------
        # 5. VALIDATION
        # --------------------------------------------------

        model.eval()

        val_loss_sum = 0.0
        val_correct = 0
        val_total = 0

        with torch.inference_mode():

            for images, labels in val_loader:

                images = images.to(
                    device,
                    non_blocking=True
                )

                labels = labels.to(
                    device,
                    non_blocking=True
                )

                outputs = model(images)

                loss = criterion(outputs, labels)

                val_loss_sum += (
                    loss.item() * images.size(0)
                )

                predictions = outputs.argmax(dim=1)

                val_correct += (
                    predictions == labels
                ).sum().item()

                val_total += labels.size(0)

        val_loss = val_loss_sum / val_total
        val_accuracy = val_correct / val_total

        scheduler.step(val_accuracy)

        history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "train_accuracy": train_accuracy,
            "val_loss": val_loss,
            "val_accuracy": val_accuracy
        })

        print(
            f"\nEpoch {epoch + 1}/{EPOCHS} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Train Acc: {train_accuracy:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Val Acc: {val_accuracy:.4f}"
        )

        # Save best validation model

        if val_accuracy > best_val_accuracy:

            best_val_accuracy = val_accuracy

            best_state = {
                key: value.detach().cpu().clone()
                for key, value in model.state_dict().items()
            }

            epochs_without_improvement = 0

            torch.save({
                "model_state_dict": best_state,
                "class_to_idx": real_train.class_to_idx,
                "seed": seed,
                "best_val_accuracy": best_val_accuracy
            }, MODEL_DIR / "best_resnet18.pth")

        else:

            epochs_without_improvement += 1

        if epochs_without_improvement >= PATIENCE:

            print("\nEarly stopping triggered.")
            break

    training_time = time.time() - start_time

    if best_state is None:
        raise RuntimeError("No valid best model was saved.")

    # Restore best validation model
    model.load_state_dict(best_state)

    # Save training history
    history_df = pd.DataFrame(history)

    history_df.to_csv(
        RESULT_DIR / "training_history.csv",
        index=False
    )

    print("\nBest validation accuracy:", best_val_accuracy)
    print(
        "Training time (minutes):",
        round(training_time / 60, 2)
    )

    # --------------------------------------------------
    # 6. TEST EVALUATION
    # --------------------------------------------------

    model.eval()

    y_true = []
    y_pred = []

    with torch.inference_mode():

        for images, labels in tqdm(
            test_loader,
            desc="Testing"
        ):

            images = images.to(
                device,
                non_blocking=True
            )

            outputs = model(images)

            predictions = outputs.argmax(dim=1)

            y_pred.extend(
                predictions.cpu().numpy()
            )

            y_true.extend(
                labels.numpy()
            )

    class_names = [
        name for name, index in sorted(
            real_train.class_to_idx.items(),
            key=lambda item: item[1]
        )
    ]

    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    macro_f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------
    # 7. SAVE METRICS
    # --------------------------------------------------

    metrics = {
        "experiment": experiment_name,
        "classifier": "ResNet18",
        "seed": seed,
        "accuracy": accuracy,
        "macro_precision": precision,
        "macro_recall": recall,
        "macro_f1": macro_f1,
        "training_time_minutes": training_time / 60,
        "real_training_images": len(real_train),
        "synthetic_training_images": len(synthetic_train),
        "combined_training_images": len(train_dataset),
        "validation_images": len(real_val),
        "test_images": len(real_test)
    }

    pd.DataFrame([metrics]).to_csv(
        RESULT_DIR / "metrics.csv",
        index=False
    )

    report = classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )

    pd.DataFrame(report).transpose().to_csv(
        RESULT_DIR / "classification_report.csv"
    )

    predictions_df = pd.DataFrame({
        "actual_class": [
            class_names[i] for i in y_true
        ],
        "predicted_class": [
            class_names[i] for i in y_pred
        ]
    })

    predictions_df.to_csv(
        RESULT_DIR / "predictions.csv",
        index=False
    )

    # --------------------------------------------------
    # 8. SAVE LOSS AND ACCURACY CURVES
    # --------------------------------------------------

    plt.figure(figsize=(8, 5))

    plt.plot(
        history_df["epoch"],
        history_df["train_loss"],
        label="Training Loss"
    )

    plt.plot(
        history_df["epoch"],
        history_df["val_loss"],
        label="Validation Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"ResNet-18 Loss - {experiment_name}")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    plt.savefig(
        RESULT_DIR / "loss_curve.png",
        dpi=300
    )

    plt.show()
    plt.close()

    plt.figure(figsize=(8, 5))

    plt.plot(
        history_df["epoch"],
        history_df["train_accuracy"],
        label="Training Accuracy"
    )

    plt.plot(
        history_df["epoch"],
        history_df["val_accuracy"],
        label="Validation Accuracy"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(f"ResNet-18 Accuracy - {experiment_name}")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    plt.savefig(
        RESULT_DIR / "accuracy_curve.png",
        dpi=300
    )

    plt.show()
    plt.close()

    # --------------------------------------------------
    # 9. CONFUSION MATRIX
    # --------------------------------------------------

    cm = confusion_matrix(
        y_true,
        y_pred
    )

    plt.figure(figsize=(8, 6))

    plt.imshow(cm, interpolation="nearest")
    plt.colorbar()

    plt.xticks(
        range(len(class_names)),
        class_names,
        rotation=45,
        ha="right"
    )

    plt.yticks(
        range(len(class_names)),
        class_names
    )

    for i in range(len(class_names)):
        for j in range(len(class_names)):
            plt.text(
                j, i,
                str(cm[i, j]),
                ha="center",
                va="center"
            )

    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")

    plt.title(
        f"Confusion Matrix - {experiment_name}"
    )

    plt.tight_layout()

    plt.savefig(
        RESULT_DIR / "confusion_matrix.png",
        dpi=300
    )

    plt.show()
    plt.close()

    # --------------------------------------------------
    # 10. SAVE EXPERIMENT CONFIGURATION
    # --------------------------------------------------

    experiment_config = {
        "experiment": experiment_name,
        "classifier": "ResNet18",
        "classifier_pretrained": "ImageNet",
        "real_training_images": len(real_train),
        "synthetic_training_images": len(synthetic_train),
        "validation_images": len(real_val),
        "test_images": len(real_test),
        "image_size": IMAGE_SIZE,
        "batch_size": BATCH_SIZE,
        "epochs_max": EPOCHS,
        "early_stopping_patience": PATIENCE,
        "seed": seed,
        "learning_rate": LEARNING_RATE,
        "weight_decay": WEIGHT_DECAY,
        "optimizer": "AdamW",
        "scheduler": "ReduceLROnPlateau",
        "best_validation_accuracy": float(best_val_accuracy),
        "test_metrics": metrics
    }

    with open(
        RESULT_DIR / "config.json",
        "w"
    ) as f:

        json.dump(
            experiment_config,
            f,
            indent=4
        )

    # --------------------------------------------------
    # 11. FINAL SUMMARY
    # --------------------------------------------------

    print("\n" + "=" * 50)
    print("FINAL RESULTS:", experiment_name)
    print("=" * 50)

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"Macro F1:  {macro_f1:.4f}")

    print("\nClassification report:")

    print(classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        zero_division=0
    ))

    print("\nResults saved in:", RESULT_DIR)

    del model
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return metrics
```
